```python
list_professores = ["Felipe", "Marcelo", "Luyz"]

print(f"Os professores de 2025 são: {list_professores}")

print(f"Em 2026, temos um novo professor e a lista vai ficar:")

novo_professor = input("informe o nome do novo professor:")

#Para ADICIONAR um novo valor fazemos
#Chamar a lista e colocar o atributo append com o novo valor
# list.append("novo valor")

list_professores.append(novo_professor)
print(list_professores)

#Para REMOVER um dos valores
# Usampos a ferramenta POP
#Chamar a lista e colocar o atributo pop, com a posição
#Do valor que queremos remover
#list.pop(index)

#Ex: Vamos remover o professor Marcelo

list_professores.pop(1)
print(list_professores)
```

```python
# Jogo 1 - Remover valor da lista
# Temos a lista
list_pets = ["meg", "duda", "pandora","joy", "garoa", "mini", "soneca", "honda", "lucky", "bingo"]

# Desafio:
# Sortear uma posição da list_pets
# Usando o comando pop, remover
```

```python
import random

list_pets = ["meg", "duda", "pandora","joy", "garoa", "mini", "soneca", "honda", "lucky", "bingo"]

sortear = random.choice(list_pets)
print(sortear)
print(f"A lista tem os elementos: {list_pets}")

posicao = int(input("Informe a posição do valor sorteado"))

print(f"O valor removido foi {list_pets[posicao]}")
list_pets.pop(posicao)


print(list_pets)
```
