```python
# Vamos jogar par ou impar

# Operadores matematicos
# + soma
# - subtracao
# / divisao
# % modulo (resto da divisao)
# * multiplicacao
# = recebe
# == igual

# Para condicionar alguma coisa, precisamos:
# Comando de condicao (if)
# Condicao da comparacao
# Resposta
# O que fazer se nao conseguir a resposta

# EX: Qual é o maior numero
n1 = int(input("informe um numero "))
n2 = int(input("informe outro numero "))

# Laço de condicao
# Se o n1 for maior que o n2 faça (faca é os dois pontos":"")

if n1 > n2:
  print("o numero 1 é maior que o numero 2")
elif n1 == n2:
  print("os valores são iguais")
else:
  print("o numero 2 é maior do que o numero 1")
```

```python
# Desafio - Criar um jogo de par ou impar

# 1 - pedir para o usuário 2 numeros
# 2 - somar estes dois numeros
# 3 - descobrir se essa soma é par ou impar
# (O desafio está aqui)
# 4 - informar se deu par ou impar
```

```python
valor1 = int(input("Informe um numero"))
valor2 = int(input("Informe outro numero"))

soma = (valor1 + valor2)

if soma % 2 == 0:
  print("(soma) é par")
else:
  print("(soma) é impar")

```
